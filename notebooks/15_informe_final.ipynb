## 1. Tablero de Control de Indicadores Clave (KPIs Nacionales)

```python
nat_hogares = df_h["fex_c"].sum()
nat_personas = df_p["fex_c"].sum()

kpis = [
 ("Acceso Total a Internet en Hogares", (df_h["acceso_internet_hogar"]*df_h["fex_c"]).sum() / nat_hogares * 100, "17.92 M hogares"),
 ("Acceso a Internet Móvil en Hogares", (df_h["tiene_internet_movil"]*df_h["fex_c"]).sum() / nat_hogares * 100, "16.91 M hogares"),
 ("Tenencia de Computador / Portátil", (df_h["tiene_computador"]*df_h["fex_c"]).sum() / nat_hogares * 100, "9.30 M hogares"),
 ("Acceso a Banda Ancha Fija", (df_h["tiene_internet_fijo"]*df_h["fex_c"]).sum() / nat_hogares * 100, "2.44 M hogares"),
 ("Uso Individual de Internet (>= 5 años)", (df_p["usa_internet"]*df_p["fex_c"]).sum() / nat_personas * 100, "40.40 M personas"),
 ("Uso Diario de Internet (sobre usuarios)", (df_p["uso_diario_internet"]*df_p["fex_c"]).sum() / (df_p["usa_internet"]*df_p["fex_c"]).sum() * 100, "Frecuencia diaria"),
]

df_kpi = pd.DataFrame(kpis, columns=["Indicador Estratégico", "Tasa Ponderada (%)", "Población Expandida / Detalle"])
df_kpi["Tasa Ponderada (%)"] = df_kpi["Tasa Ponderada (%)"].map(lambda x: f"{x:.1f}%")

print("TABLERO EJECUTIVO DE KPIS — COLOMBIA ECV 2024")
print("=" * 75)
display(df_kpi.to_string(index=False))
```
