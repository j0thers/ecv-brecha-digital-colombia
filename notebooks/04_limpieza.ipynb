```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
from pathlib import Path

from src.config import PATHS, RAW_SOURCES, CSV_PARAMS
from src.cleaning import clean_hogar, clean_tic, clean_educacion, clean_diseno
```

## 4.1 Carga y Limpieza del Módulo Hogar (Condiciones de Vida)

```python
df_hogar_raw = pd.read_csv(RAW_SOURCES['hogar'], **CSV_PARAMS)
df_hogar_clean = clean_hogar(df_hogar_raw)

# Guardar en data/processed/
out_hogar = PATHS['processed'] / 'hogar_clean.parquet'
df_hogar_clean.to_parquet(out_hogar, index=False)
print(f'[OK] Guardado: {out_hogar.name} ({df_hogar_clean.shape[0]:,} filas x {df_hogar_clean.shape[1]} columnas)')
```
