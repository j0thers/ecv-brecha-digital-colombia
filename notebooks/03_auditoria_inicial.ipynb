## 3.5 Correspondencia entre módulos

```python
# ¿Los hogares en TIC coinciden con los de Condiciones de vida?
hogares_hogar = set(df_hogar[ID_HOGAR].apply(tuple, axis=1))
hogares_tic = set(df_tic[ID_HOGAR].apply(tuple, axis=1))

print(f'Hogares en módulo Hogar : {len(hogares_hogar):,}')
print(f'Hogares en módulo TIC : {len(hogares_tic):,}')
print(f'Intersección : {len(hogares_hogar & hogares_tic):,}')
print(f'Solo en Hogar : {len(hogares_hogar - hogares_tic):,}')
print(f'Solo en TIC : {len(hogares_tic - hogares_hogar):,}')
```
